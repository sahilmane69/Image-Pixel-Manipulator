```python
img = np.zeros((200, 200, 3), dtype=np.uint8)

plt.imshow(img)
plt.title("Image")
plt.axis("off")
plt.show()
```

```python
img[10:60, 20:90] = np.full((50, 70, 3), [255, 0, 0], dtype=np.uint8)
img[70:120, 20:90] = np.full((50, 70, 3), [0, 255, 0], dtype=np.uint8)
img[130:180, 20:90] = np.full((50, 70, 3), [0, 0, 255], dtype=np.uint8)

img[20:180, 110:125] = np.full((160, 15, 3), [255, 255, 0], dtype=np.uint8)
img[20:180, 135:150] = np.full((160, 15, 3), [255, 0, 255], dtype=np.uint8)
img[20:180, 160:175] = np.full((160, 15, 3), [0, 255, 255], dtype=np.uint8)

plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
crop = img[40:160, 40:160]
flip = img[:, ::-1]

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(crop)
plt.title("Crop")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(flip)
plt.title("Flip")
plt.axis("off")

plt.show()
```

```python
red = img[:, :, 0]
green = img[:, :, 1]
blue = img[:, :, 2]

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(red, cmap="gray")
plt.title("Red")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(green, cmap="gray")
plt.title("Green")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blue, cmap="gray")
plt.title("Blue")
plt.axis("off")

plt.show()
```

```python
bright = img.astype(np.float32) + 60
bright = np.clip(bright, 0, 255).astype(np.uint8)

dark = img.astype(np.float32) - 60
dark = np.clip(dark, 0, 255).astype(np.uint8)

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(bright)
plt.title("Bright")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(dark)
plt.title("Dark")
plt.axis("off")

plt.show()
```

```python
gray = img.mean(axis=2)
mask = gray > 100

plt.imshow(mask, cmap="gray")
plt.title("Bright Pixels")
plt.axis("off")
plt.show()
```

```python
uint8_img = np.zeros((200, 200, 3), dtype=np.uint8)
float64_img = np.zeros((200, 200, 3), dtype=np.float64)

print("uint8 memory:", uint8_img.nbytes, "bytes")
print("float64 memory:", float64_img.nbytes, "bytes")
```

```python
highlight = img.copy()

gray = img.mean(axis=2)
mask = gray > 100

highlight[mask] = [255, 255, 255]

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(highlight)
plt.title("Bright Area Highlighted")
plt.axis("off")

plt.show()
```
