```python
img = np.zeros((200, 200, 3), dtype=np.uint8)

plt.imshow(img)
plt.title("Image")
plt.axis("off")
plt.show()
```

```python

img[10:60, 20:90] = [255, 0, 0]       
img[70:120, 30:150] = [0, 255, 0]     
img[130:180, 50:180] = [0, 0, 255]    

plt.imshow(img)
plt.title("RGB Color")
plt.axis("off")
plt.show()
```

```python
img[:, 0:20] = [255, 255, 0]       
img[:, 180:200] = [255, 0, 255]    
img[0:20, :] = [0, 255, 255]       

plt.imshow(img)
plt.title("Image")
plt.axis("off")
plt.show()
```

```python
cropped = img[50:150, 50:150]

plt.imshow(cropped)
plt.title("Cropped Image")
plt.axis("off")
plt.show()
```

```python
horizontal_flip = img[:, ::-1]
vertical_flip = img[::-1, :]

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(horizontal_flip)
plt.title("Horizontal Flip")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(vertical_flip)
plt.title("Vertical Flip")
plt.axis("off")

plt.show()
```
